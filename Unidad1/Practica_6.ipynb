```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import math

df = pd.read_csv("autos2.csv")
print(df.head())
print("\nInformación del dataset:")
print(df.info())

precios = df['price'].dropna()

# Valor mínimo
print("\nValor mínimo de precios:", precios.min())

# Valor máximo
print("Valor máximo de precios:", precios.max())

# Moda
print("Moda de precios:", precios.mode().tolist())

# Media
print("Media de precios:", precios.mean())

# Mediana
print("Mediana de precios:", precios.median())

# Desviación estándar
print("Desviación estándar:", precios.std())

# Estadísticos y cuartiles
print("\nEstadísticos descriptivos:")
print(precios.describe())
```

```python
# REGLA DE STURGES
n = len(precios)

K = math.ceil(1 + math.log2(n))

print("\nNúmero de datos:", n)
print("Número óptimo de bins según Sturges:", K)
```

```python
# HISTOGRAMA DE PRECIOS
plt.figure(figsize=(10, 6))

plt.hist(precios, bins=K, edgecolor='black')

plt.title('Histograma de precios de automóviles')
plt.xlabel('Precio del automóvil ($)')
plt.ylabel('Frecuencia')

plt.grid(axis='y', alpha=0.3)

plt.show()
```

```python
# DIAGRAMA DE CAJA Y BIGOTES:
# PRECIO Y TIPO

plt.figure(figsize=(10, 6))

df.boxplot(
    column='price',
    by='body-style',
    grid=False
)

plt.title('Precio de automóviles según tipo de vehículo')
plt.suptitle('')  # Elimina el título automático
plt.xlabel('Tipo de vehículo')
plt.ylabel('Precio ($)')

plt.xticks(rotation=0)

plt.show()
```

```python
# DIAGRAMA DE CAJA Y BIGOTES:
# PRECIO Y TRACCION

plt.figure(figsize=(10, 6))

df.boxplot(
    column='price',
    by='drive-wheels',
    grid=False
)

plt.title('Precio de automóviles según tipo de tracción')
plt.suptitle('')
plt.xlabel('Tipo de tracción')
plt.ylabel('Precio ($)')

plt.show()
```

```python
# DIAGRAMA DE CAJA Y BIGOTES:
# PRECIO Y UBICACION
plt.figure(figsize=(8, 6))

df.boxplot(
    column='price',
    by='engine-location',
    grid=False
)

plt.title('Precio de automóviles según ubicación del motor')
plt.suptitle('')
plt.xlabel('Ubicación del motor')
plt.ylabel('Precio ($)')

plt.show()
```

4. El **tipo de vehículo** presenta diferencias importantes en el precio. Los hatchbacks son mas baratos, mientras que hardtops y convertibles presentan precios más altos.
