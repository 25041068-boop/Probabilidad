```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error
from scipy.stats import linregress, pearsonr

ARCHIVOS = {
    "Grupo 1": "grupo1.csv",
    "Grupo 2": "grupo2.csv",
    "Grupo 3": "grupo3.csv"
}
CARPETA_SALIDA = "resultados_practica_unidad5"
os.makedirs(CARPETA_SALIDA, exist_ok=True)

def fuerza_correlacion(r):
    a = abs(r)
    if a >= 0.80:
        fuerza = "muy fuerte"
    elif a >= 0.60:
        fuerza = "fuerte"
    elif a >= 0.40:
        fuerza = "moderada"
    elif a >= 0.20:
        fuerza = "débil"
    else:
        fuerza = "muy débil"
    direccion = "positiva" if r > 0 else ("negativa" if r < 0 else "nula")
    return fuerza, direccion

resumen = []

for nombre, archivo in ARCHIVOS.items():
    print("\n" + "=" * 75)
    print(nombre.upper())
    print("=" * 75)

    # 1. Carga y exploracion
    df = pd.read_csv(archivo)
    columnas_requeridas = ["HorasEstudio", "Calificacion"]
    if not all(col in df.columns for col in columnas_requeridas):
        raise ValueError(f"{archivo} debe contener las columnas {columnas_requeridas}")

    df = df[columnas_requeridas].copy()
    df["HorasEstudio"] = pd.to_numeric(df["HorasEstudio"], errors="coerce")
    df["Calificacion"] = pd.to_numeric(df["Calificacion"], errors="coerce")
    df = df.dropna()

    print("\nPrimeros cinco registros:")
    print(df.head())
    print("\nDescripcion estadistica:")
    print(df.describe())
    print(f"\nCantidad de registros validos: {len(df)}")

    X = df[["HorasEstudio"]]
    y = df["Calificacion"]

    # 2. Regresion lineal
    modelo = LinearRegression()
    modelo.fit(X, y)
    pendiente = float(modelo.coef_[0])
    intercepto = float(modelo.intercept_)
    predicciones = modelo.predict(X)
    r2 = r2_score(y, predicciones)
    rmse = np.sqrt(mean_squared_error(y, predicciones))

    # 3. Correlacion de Pearson y prueba de significancia
    r, p_pearson = pearsonr(df["HorasEstudio"], y)
    prueba = linregress(df["HorasEstudio"], y)
    p_valor = float(prueba.pvalue)
    fuerza, direccion = fuerza_correlacion(r)
    significativo = p_valor < 0.05

    print("\nMODELO DE REGRESION LINEAL")
    print(f"Ecuacion: Calificacion = {pendiente:.4f} * HorasEstudio + ({intercepto:.4f})")
    print(f"Pendiente: {pendiente:.4f}")
    print(f"Intercepto: {intercepto:.4f}")
    print(f"R cuadrada: {r2:.4f} ({r2*100:.2f}% de variacion explicada)")
    print(f"RMSE: {rmse:.4f} puntos de calificacion")

    print("\nCORRELACION DE PEARSON")
    print(f"r = {r:.4f}")
    print(f"p-valor = {p_valor:.8g}")
    print(f"La relacion es {direccion} y {fuerza}.")

    # 4. Respuestas interpretativas
    print("\nINTERPRETACION Y RESPUESTAS")
    print("1. Coeficiente de regresion (pendiente):")
    if pendiente > 0:
        print(f"   Por cada hora adicional de estudio, la calificacion estimada aumenta "
              f"{pendiente:.2f} puntos, en promedio.")
    elif pendiente < 0:
        print(f"   Por cada hora adicional de estudio, la calificacion estimada disminuye "
              f"{abs(pendiente):.2f} puntos, en promedio.")
    else:
        print("   La recta no estima cambios en la calificacion por hora adicional.")

    print("2. Coeficiente de correlacion:")
    print(f"   r={r:.4f} indica una relacion {direccion} y {fuerza} entre las variables.")

    print("3. Significancia estadistica:")
    if significativo:
        print(f"   Si. p={p_valor:.4g} es menor que 0.05; se rechaza la hipotesis "
              "de pendiente igual a cero. Hay evidencia estadistica de relacion lineal.")
    else:
        print(f"   No. p={p_valor:.4f} es mayor o igual que 0.05; no hay evidencia "
              "suficiente para afirmar que la pendiente sea distinta de cero.")

    print("4. Confiabilidad para predecir:")
    print(f"   R²={r2:.4f}: el modelo explica {r2*100:.2f}% de la variabilidad observada "
          f"en las calificaciones. El error cuadratico medio de raiz es {rmse:.2f} puntos.")
    if r2 >= 0.80:
        print("   El ajuste dentro del rango observado es alto; aun asi, las predicciones "
              "son estimaciones y no deben extrapolarse fuera del rango de horas.")
    elif r2 >= 0.50:
        print("   El ajuste es moderado; puede servir como referencia, pero conserva "
              "una parte importante de variabilidad sin explicar.")
    else:
        print("   El ajuste es bajo; no es confiable usar solo las horas de estudio "
              "para predecir las calificaciones.")

    # 5. Histograma de calificaciones
    plt.figure(figsize=(8, 5))
    plt.hist(df["Calificacion"], bins=10, edgecolor="black")
    plt.title(f"Histograma de calificaciones - {nombre}")
    plt.xlabel("Calificacion final")
    plt.ylabel("Frecuencia")
    plt.grid(axis="y", alpha=0.25)
    plt.tight_layout()
    plt.savefig(os.path.join(CARPETA_SALIDA, f"{nombre.lower().replace(' ', '')}_histograma.png"), dpi=150)
    plt.close()

    # 6. Diagrama de dispersion y linea de regresion
    orden = np.argsort(df["HorasEstudio"].to_numpy())
    x_ordenado = df["HorasEstudio"].to_numpy()[orden]
    y_ajustado = predicciones[orden]

    plt.figure(figsize=(8, 5))
    plt.scatter(df["HorasEstudio"], y, alpha=0.8, label="Datos observados")
    plt.plot(x_ordenado, y_ajustado, linewidth=2, label="Linea de regresion")
    plt.title(f"Horas de estudio y calificaciones - {nombre}")
    plt.xlabel("Horas de estudio")
    plt.ylabel("Calificacion final")
    plt.legend()
    plt.grid(alpha=0.25)
    plt.tight_layout()
    plt.savefig(os.path.join(CARPETA_SALIDA, f"{nombre.lower().replace(' ', '')}_regresion.png"), dpi=150)
    plt.close()

    resumen.append({
        "Grupo": nombre,
        "Registros": len(df),
        "Pendiente": pendiente,
        "Intercepto": intercepto,
        "Pearson_r": r,
        "p_valor": p_valor,
        "R2": r2,
        "RMSE": rmse,
        "Significativo_0.05": "Si" if significativo else "No"
    })

# 7. Tabla comparativa y conclusion general
tabla = pd.DataFrame(resumen)
tabla.to_csv(os.path.join(CARPETA_SALIDA, "resumen_comparativo.csv"), index=False)
print("\n" + "=" * 75)
print("COMPARACION FINAL DE LOS TRES GRUPOS")
print("=" * 75)
print(tabla.to_string(index=False, float_format=lambda x: f"{x:.4f}"))

print("\nCONCLUSION GENERAL")
for fila in resumen:
    print(f"- {fila['Grupo']}: r={fila['Pearson_r']:.4f}, R²={fila['R2']:.4f}, "
          f"p={fila['p_valor']:.4g}.")
print("La comparacion debe considerar la direccion y fuerza de r, la pendiente, "
      "la significancia (p<0.05) y el porcentaje de variacion explicado (R²).")
print("La asociacion estadistica no demuestra por si sola causalidad. Otros factores "
      "tambien pueden influir en el rendimiento academico.")
print(f"\nGraficas y tabla guardadas en: {CARPETA_SALIDA}")
```
