```python

import matplotlib.pyplot as plt

x = [0, 10, 20, 50]
p = [0.50, 0.30, 0.15, 0.05]

print("Suma de probabilidades:", sum(p))

media = sum(xi * pi for xi, pi in zip(x, p))
print("Valor esperado:", media)

ex2 = sum(xi**2 * pi for xi, pi in zip(x, p))

varianza = ex2 - media**2
desviacion = varianza**0.5

print("Varianza:", varianza)
print("Desviación estándar:", desviacion)

# Gráfica de distribución de probabilidad
plt.figure()
plt.stem(x, p)
plt.title("Distribución de probabilidad - Ejercicio 2")
plt.xlabel("Monto del premio (pesos)")
plt.ylabel("Probabilidad")
plt.grid(True)
plt.show()

# Calcular función acumulada
f = []
acumulada = 0

for pi in p:
    acumulada += pi
    f.append(acumulada)

# Gráfica de distribución acumulada
plt.figure()
plt.step(x, f, where="post")
plt.scatter(x, f)
plt.title("Función acumulada - Ejercicio 2")
plt.xlabel("Monto del premio (pesos)")
plt.ylabel("F(x)")
plt.ylim(0, 1.1)
plt.grid(True)
plt.show()
```
