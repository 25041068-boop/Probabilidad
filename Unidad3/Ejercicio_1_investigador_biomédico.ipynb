```python

import matplotlib.pyplot as plt

x = [0, 1, 2, 3, 4, 5]
p = [0.10, 0.25, 0.30, 0.20, 0.10, 0.05]

print("Suma de probabilidades:", sum(p))

media = sum(xi * pi for xi, pi in zip(x, p))
print("Valor esperado:", media)

ex2 = sum(xi**2 * pi for xi, pi in zip(x, p))

varianza = ex2 - media**2
desviacion = varianza**0.5

print("Varianza:", varianza)
print("Desviación estándar:", desviacion)

# Distribución de probabilidad
plt.figure()
plt.stem(x, p)
plt.title("Distribución de probabilidad - Ejercicio 1")
plt.xlabel("Número de células anormales")
plt.ylabel("Probabilidad")
plt.grid(True)
plt.show()

# Función de distribución acumulada
f = []
acumulada = 0

for pi in p:
    acumulada += pi
    f.append(acumulada)

plt.figure()
plt.step(x, f, where="post")
plt.scatter(x, f)
plt.title("Función de distribución acumulada - Ejercicio 1")
plt.xlabel("Número de células anormales")
plt.ylabel("F(x)")
plt.ylim(0, 1.1)
plt.grid(True)
plt.show()
```
