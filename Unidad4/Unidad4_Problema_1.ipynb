```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

media = 70
desv_std = 4
num_datos = 500

z1 = (68 - media) / desv_std
z2 = (72 - media) / desv_std

probabilidad = norm.cdf(z2) - norm.cdf(z1)

estudiantes = probabilidad * num_datos

print("Probabilidad:", probabilidad)
print("Porcentaje:", probabilidad * 100, "%")
print("Numero de estudiantes:", estudiantes)

# Generar datos para la distribucion normal
datos = np.random.normal(media, desv_std, num_datos)

# Histograma de los pesos
plt.hist(datos, bins=20, density=True, alpha=0.6,
         label="Pesos de los estudiantes")

# Funcion de densidad de probabilidad
x = np.linspace(55, 85, 500)
y = norm.pdf(x, media, desv_std)
plt.plot(x, y, label="Distribucion normal")

# Marcar los limites de 68 y 72 kg
plt.axvline(68, linestyle="--", label="68 kg")
plt.axvline(72, linestyle="--", label="72 kg")

plt.title("Distribucion normal de los pesos")
plt.xlabel("Peso en kilogramos")
plt.ylabel("Densidad de probabilidad")
plt.legend()
plt.grid()
plt.show()
```
