```python
import math

# PROBLEMA 2: DISTRIBUCION BINOMIAL
# n = 1000 motores; p = 0.001 probabilidad de defecto.
# Se usa Binomial porque cada motor puede ser defectuoso o no,
# con probabilidad de defecto constante en cada ensayo.
# Poisson se incluye solo como aproximacion, con lambda = n * p.

def binomial(n, k, p):
    """Calcula la probabilidad P(X = k) con distribucion binomial."""
    return math.comb(n, k) * (p**k) * ((1 - p)**(n - k))


def poisson(lam, k):
    """Calcula la probabilidad P(X = k) con distribucion de Poisson."""
    return math.exp(-lam) * lam**k / math.factorial(k)


n = 1000
p = 0.001
lam = n * p

print("Problema 2 - Distribucion Binomial")
print(f"Numero de motores (n) = {n}")
print(f"Probabilidad de defecto (p) = {p}")
print(f"Probabilidad de no defecto (1-p) = {1 - p}")
print(f"Lambda para la aproximacion de Poisson = {lam}")

for k, descripcion in [
    (0, "ningun motor defectuoso"),
    (3, "exactamente 3 motores defectuosos"),
    (4, "exactamente 4 motores defectuosos"),
]:
    p_binomial = binomial(n, k, p)
    p_poisson = poisson(lam, k)

    print(f"\nProbabilidad de {descripcion}:")
    print(f"Binomial exacta = {p_binomial:.8f} ({p_binomial * 100:.4f}%)")
    print(f"Aproximacion Poisson = {p_poisson:.8f} ({p_poisson * 100:.4f}%)")
```
