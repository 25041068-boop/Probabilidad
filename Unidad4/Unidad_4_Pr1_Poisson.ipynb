```python
import math

# PROBLEMA 1: DISTRIBUCION DE POISSON
# Promedio: 5 ingresos a cuidados intensivos por dia.
# Se usa Poisson porque se cuentan eventos en un intervalo de tiempo
# y se conoce el promedio de eventos por intervalo.

def poisson(lam, k):
    """Calcula la probabilidad P(X = k) con distribucion de Poisson."""
    return math.exp(-lam) * lam**k / math.factorial(k)


lam = 5

# a) Probabilidad de que ingresen exactamente 2 personas.
p_dos = poisson(lam, 2)
print("Problema 1 - a) Exactamente 2 personas")
print(f"P(X = 2) = {p_dos:.8f}")
print(f"Porcentaje = {p_dos * 100:.4f}%")

# b) Probabilidad de que ingresen mas de 10 personas.
# P(X > 10) = 1 - P(X <= 10)
p_hasta_diez = sum(poisson(lam, k) for k in range(0, 11))
p_mas_de_diez = 1 - p_hasta_diez

print("\nProblema 1 - b) Mas de 10 personas")
print(f"P(X > 10) = {p_mas_de_diez:.8f}")
print(f"Porcentaje = {p_mas_de_diez * 100:.4f}%")
```
